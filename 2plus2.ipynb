````python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
````

Нам приедтся работать с двоичным кодом: 

* 2 — это 10 
* 4 — это 100

Мы собираемся посчитать 2 + 2 = 4, то есть 10 + 10 = 100 

Каждый кубит умеет хранить всего одну цифру двоичной записи. Поэтому для 10 + 10 = 100 нам нужно как минимум 7 кубитов. Один нужен про запас, потом обяъсню его смысл. 

Чтобы записать ответ 100 и показать его, нужно еще 3 обычных бита

Так что создаем квантовый процесс на 8 кубитов и 3 бита

````python
qc = QuantumCircuit(8, 3)
````

Мы уже создали 8 кубитов, по умолчанию они все равны 0, то есть сейчас внутри нашего компьютера лежит `00 00 000 0`: 

* Два первых нолика для первой 2 (на двоичном коде 10)
* Два вторых нолика для второй 2 (на двоичном коде 10)
* Потом три нолика для 4 (на двоичном коде 100)
* Один последний про запас

Записываем в него две двойки, которые сбираемся складывать, получается: `10 10 000 0` (заменили нолик № 1 и № 3 на единички)

````python
qc.x(1) # 00 00 000 0 → 10 00 000 0
qc.x(3) # 00 00 000 0 → 10 10 000 0
````

Теперь начинаем прям в столбик складывать 10 и 10 на квантовом компьютере. Сначала складывем правый столбик 

```
  10
+ 10
----
   0
```

````python
qc.cx(0, 4)
qc.cx(2, 4)
qc.ccx(0, 2, 6) 
````

По итогу все еще имеем `10 10 000 0`, так как в правом столбике вышел ноль. 

Теперь считаем средний столбик. Единичку нам надо записать в следующий разряд (надеюсь, все помнят, как считать в столбик). Для этого и нужен 3 кубит — в немго мы записываем тот факт, что что-то надо перенести в следующий разряд

```
  10
+ 10
----
  00
```

````python
qc.cx(1, 5)
qc.cx(3, 5)
qc.cx(6, 5)
````

Теперь имеем `10 10 000 1` — переносящийся разряд мы записали в отдельный кубит

Теперь считаем левый столбик, последний шаг. 

```
  10
+ 10
----
 100
```

````python
qc.ccx(1, 3, 7)
qc.ccx(1, 6, 7)
qc.ccx(3, 6, 7)
````

В теории получилось `10 10 100 0`. Чтобы проверить это, переписываем результат в биты, уничтожаем кубиты и смотрим на ответ, рисуя его 

````python
qc.measure(4, 0)
qc.measure(5, 1)
qc.measure(7, 2)
qc.draw(output = "mpl") 
````
