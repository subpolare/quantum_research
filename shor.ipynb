```python
from pathlib import Path
from qiskit_ibm_runtime import QiskitRuntimeService
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.backends import default_backend
```

```python
QiskitRuntimeService.save_account(
    channel        = 'ibm_quantum_platform',
    token          = 'YOUR_API_KEY',
    instance       = 'YOUR_CRN',
    set_as_default = True,
    overwrite      = True,
)
```

```python
PUBLIC_KEY_PATH = Path('/Users/subpolare/.ssh/id_rsa.pub')
key_data = PUBLIC_KEY_PATH.read_bytes()

try:
    public_key = serialization.load_ssh_public_key(
        key_data,
        backend = default_backend(),
    )
except ValueError:
    public_key = serialization.load_pem_public_key(
        key_data,
        backend = default_backend(),
    )

numbers = public_key.public_numbers()

N = numbers.n
print('Публичный ключ (modulus N):', N)

num_target = N.bit_length()
num_control = 2 * num_target
min_logical_qubits = num_target + num_control
print(f'Нужно минимум: {min_logical_qubits} кубитов')

service = QiskitRuntimeService()
backend = service.least_busy(
    simulator      = False,
    operational    = True,
    min_num_qubits = min_logical_qubits,
)

print('Found backend:', backend.name)
```
